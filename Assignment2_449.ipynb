```python
roll_no = "1024170449" 
L = [int(digit) * 10 for digit in roll_no]

print("Original list:", L)

L.append(100) 
print("After append:", L)

L.insert(2, 55) 
print("After insert:", L)

L.remove(55) 
print("After remove:", L)

L.pop() 
print("After pop:", L)

L.sort()
print("Sorted ascending:", L)

L.sort(reverse=True)
print("Sorted descending:", L)

print("First three:", L[:3], "| Last three:", L[-3:])

average = sum(L) / len(L)
greater_than_avg = [x for x in L if x > average]
print("Elements greater than average:", greater_than_avg)
```

```python
roll_no = "1024170449"
L = [int(digit) * 10 for digit in roll_no]
scores = tuple(L[:8])

highest = max(scores)
highest_idx = scores.index(highest)
lowest = min(scores)
lowest_count = scores.count(lowest)
print("Highest:", highest, "at index", highest_idx)
print("Lowest:", lowest, "appears", lowest_count, "times")

reversed_list = list(scores[::-1])
print("Reversed as list:", reversed_list)

user_score = int(input("Enter a score: ")) 
if user_score in scores:
    print("First occurrence at index:", scores.index(user_score))
else:
    print("Score not present")

try:
    scores[0] = 100
except TypeError as e:
    print("Error when modifying tuple:", e)

first, second, *remaining = scores
print("First:", first, "| Second:", second, "| Remaining:", remaining)
```

```python
import random

roll_no_int = 1024170449
random.seed(roll_no_int)

rand_list = [random.randint(100, 900) for _ in range(100)]

odds = [n for n in rand_list if n % 2 != 0]
print("Total odd numbers:", len(odds))

evens = [n for n in rand_list if n % 2 == 0]
print("Total even numbers:", len(evens))

def is_prime(num):
    if num <= 1: return False
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0: return False
    return True

primes = [n for n in rand_list if is_prime(n)]
print("Total prime numbers:", len(primes))
print("List of primes:", primes)

most_frequent = max(set(rand_list), key=rand_list.count)
count = rand_list.count(most_frequent)
print("Most frequent number:", most_frequent, "(appears", count, "times)")
```

```python
roll_no = "10241470449"
digits = [int(d) for d in roll_no[:8]]

A = {d * 7 for d in digits}
B = {d * 9 for d in digits}
print("Set A:", A)
print("Set B:", B)

print("Union:", A | B)
print("Intersection:", A & B)
print("A - B:", A - B)
print("B - A:", B - A)
print("Symmetric difference:", A ^ B)
print("Is A a subset of B?:", A.issubset(B))
print("Is B a superset of A?:", B.issuperset(A))

x = int(input("Enter value to remove from A: "))
A.discard(x)
print("Set A after removal:", A)
```
